```python
import pandas as pd

df = pd.read_csv("Dataset/HR_Analytics.csv")

print("Shape of Dataset:", df.shape)
print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
print(df['Attrition'].value_counts())

print("\nAttrition Percentage:")
print(df['Attrition'].value_counts(normalize=True) * 100)
```

```python
department_attrition = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100

print(department_attrition.round(2))
```

```python
overtime_attrition = pd.crosstab(
    df['OverTime'],
    df['Attrition'],
    normalize='index'
) * 100

print(overtime_attrition.round(2))
```

```python
job_sat_attrition = pd.crosstab(
    df['JobSatisfaction'],
    df['Attrition'],
    normalize='index'
) * 100

print(job_sat_attrition.round(2))
```

```python
numeric_df = df.select_dtypes(include='number')

correlation_matrix = numeric_df.corr()

print(correlation_matrix.round(2))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix – HR Analytics')
plt.tight_layout()
plt.show()
```

```python
Q1 = df['MonthlyIncome'].quantile(0.25)
Q3 = df['MonthlyIncome'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['MonthlyIncome'] < lower_bound) |
    (df['MonthlyIncome'] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

# ATTRITION ANALYSIS

```python
# plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Attrition')

plt.title('Employee Attrition Distribution')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')

plt.tight_layout()
plt.savefig("Visualizations/attrition_distribution.png",
            dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

# Department-wise Attrition

```python
# plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='Department',
    hue='Attrition'
)

plt.title('Department-wise Employee Attrition')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')

plt.tight_layout()
plt.savefig("Visualizations/department_attrition.png",
            dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

# Overtime vs Attrition

```python
# plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition'
)

plt.title('Overtime vs Employee Attrition')
plt.xlabel('Overtime')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')

plt.tight_layout()
plt.savefig("Visualizations/overtime_attrition.png",
            dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

# Job Satisfaction vs Attrition

```python
# plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x='JobSatisfaction',
    hue='Attrition'
)

plt.title('Job Satisfaction Level vs Employee Attrition')
plt.xlabel('Job Satisfaction Level')
plt.ylabel('Number of Employees')
plt.legend(title='Attrition')

plt.tight_layout()
plt.savefig("Visualizations/job_satisfaction_attrition.png",
            dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

# Monthly Income by Job Role

```python
role_income = (
    df.groupby('JobRole')['MonthlyIncome']
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

# plt.figure(figsize=(10, 6))

sns.barplot(
    data=role_income,
    x='MonthlyIncome',
    y='JobRole'
)

plt.title('Average Monthly Income by Job Role')
plt.xlabel('Average Monthly Income')
plt.ylabel('Job Role')

plt.tight_layout()
plt.savefig("Visualizations/income_by_job_role.png",
            dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```

# Age Distribution

```python
# plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='Age',
    bins=15,
    kde=True
)

plt.title('Employee Age Distribution')
plt.xlabel('Age')
plt.ylabel('Number of Employees')

plt.tight_layout()
plt.savefig("Visualizations/age_distribution.png",
            dpi=300, bbox_inches="tight")
plt.show()
plt.close()
```
